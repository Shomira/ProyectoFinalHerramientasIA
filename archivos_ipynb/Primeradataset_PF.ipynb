```python
from faker import Faker
import pandas as pd
import random

fake = Faker()
```

```python


ciudades_ecuador = ["Quito", "Guayaquil", "Cuenca", "Loja", "Baños", "Manta", "Otavalo", "Riobamba", "Zamora", "Pastaza"]
tipos_viaje = ["Vacaciones", "Negocios", "Estudio", "Aventura", "Visita Familia"]
transportes = ["Avión", "Bus", "Auto", "Motociclista", "Ciclista"]
hospedaje = ["Hotel", "Hostal", "Airbnb", "Resort", "Casa huespedes"]

data = []

for _ in range(1000):
    personas = random.randint(1, 6) # Numero de personas que viajan puede ser entre 1 a 6
    dias = random.randint(1, 15) # dias que durqa el viaje, entre 1 y 15

    pasaje = random.randint(100, 1200) # Costo del pasaje (ida o ida/vuelta), entre 100 y 1200 dólares
    #hotel = dias * random.randint(20, 120) # Costo del hotel - Se multiplica: días de estadía × costo diario del hotel -El hotel puede costar entre 20 y 120 por día
    comida = dias * personas * random.randint(10, 40) # se multiplica los dias por el numero de personas y e l  costo diario por persona (10 a 40)
    extras = random.randint(20, 300) # Souvenirs, tpago guias locales, tours, transporte, entrada a los establecimientos. de 20 a 300
    #fecha_viaje = fake.date_between(start_date='-1y', end_date='today')
 #   total = pasaje + comida + extras # Suma de todos los gasstos

    data.append({
        "edad": random.randint(18, 70),
        "pais_origen": fake.country(),
        "ciudad_ecuador": random.choice(ciudades_ecuador),
        "tipo_viaje": random.choice(tipos_viaje),
        "fecha_viaje":  fake.date_between(start_date='-1y', end_date='today'),
        "transporte": random.choice(transportes),
        "hospedaje": random.choice(hospedaje),
        "personas_viaje": personas,
        "dias_estancia": dias,
        "pasajes": pasaje,
       # "costo_hotel": hotel,
        "costot_comida": comida,
        "gastos_extras": extras,
       # "gasto_total": total,
        "satisfaccion": random.randint(1, 10)
    })

df = pd.DataFrame(data)

df.head()
```

```python
df["fecha_viaje"] = pd.to_datetime(df["fecha_viaje"]) # Pandas convirtio en tipo object, se transformo a tipo de dato datetime
```
