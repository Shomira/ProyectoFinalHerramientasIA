```python
import sqlite3

comp = sqlite3.connect("/content/sample_data/base_comp_turismo.db")
cursor = comp.cursor()

# ============= TABLA DETALLE CIUDADES =============
cursor.execute("""
CREATE TABLE IF NOT EXISTS detalle_ciudades (
    id_ciudad INTEGER PRIMARY KEY AUTOINCREMENT,
    ciudad_ecuador TEXT UNIQUE,
    provincia TEXT,
    region TEXT,
    clima TEXT,
    altitud INTEGER,
    atractivo_principal TEXT
)
""")

# ============= HOSPEDAJE =============
cursor.execute("""
CREATE TABLE IF NOT EXISTS detalle_hospedaje (
    id_hospedaje INTEGER PRIMARY KEY AUTOINCREMENT,
    hospedaje TEXT UNIQUE,
    categoria TEXT,
    precio_noche_promedio INTEGER,
    incluye_desayuno TEXT
)
""")

cursor.executemany("""
INSERT OR IGNORE INTO detalle_hospedaje
(hospedaje, categoria, precio_noche_promedio, incluye_desayuno)
VALUES (?, ?, ?, ?)
""", [
    ("Hotel", "4 estrellas", 90, "Sí"),
    ("Hostal", "Económico", 25, "No"),
    ("Airbnb", "Privado", 50, "Depende"),
    ("Resort", "Lujo", 250, "Sí"),
    ("Casa huespedes", "Familiar", 35, "No")
])


# ======== NSERTAR INFORMACIÓN COMPLEMENTARIA =============


# CIUDADES
cursor.executemany("""
INSERT OR IGNORE INTO detalle_ciudades
(ciudad_ecuador, provincia, region, clima, altitud, atractivo_principal)
VALUES (?, ?, ?, ?, ?, ?)
""", [
    ("Quito", "Pichincha", "Sierra", "Frío", 2850, "Centro Histórico"),
    ("Guayaquil", "Guayas", "Costa", "Cálido", 4, "Malecón 2000"),
    ("Cuenca", "Azuay", "Sierra", "Templado", 2560, "Arquitectura Colonial"),
    ("Loja", "Loja", "Sierra", "Templado", 2100, "Turismo Cultural"),
    ("Baños", "Tungurahua", "Sierra", "Húmedo", 1820, "Cascadas"),
    ("Manta", "Manabí", "Costa", "Cálido", 6, "Playas"),
    ("Otavalo", "Imbabura", "Sierra", "Frío", 2530, "Mercado Artesanal"),
    ("Riobamba", "Chimborazo", "Sierra", "Frío", 2750, "Volcán Chimborazo"),
    ("Zamora", "Zamora Chinchipe", "Amazonía", "Tropical", 970, "Naturaleza"),
    ("Pastaza", "Pastaza", "Amazonía", "Húmedo", 950, "Selva Amazónica")
])

comp.commit()

# ============= VALIDACION =============

print("TABLA CIUDADES")
for row in cursor.execute("SELECT * FROM detalle_ciudades"):
    print(row)

print("\nDEATLLE HOSPEDAJE")
for row in cursor.execute("SELECT * FROM detalle_hospedaje"):
    print(row)

comp.close()
```
